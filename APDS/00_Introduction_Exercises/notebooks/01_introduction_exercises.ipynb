## Step 0: Portable Environment Bootstrap (Local IDE / Jupyter + Google Colab)

Locates the `APDS/00_Introduction_Exercises/` project folder (anchored on
`README.md`) by walking upward from the current working directory — so the
notebook runs identically from the repository root, the project folder or
its `notebooks/` sub-folder. On Google Colab the repository is cloned into
`/content/APSD_TAI_LAB` automatically.


```python
import os
import subprocess
import sys
from pathlib import Path

PROJECT_NAME = "00_Introduction_Exercises"
ANCHOR_REL = "README.md"  # file that uniquely identifies the project folder
REPO_URL = "https://github.com/Vtheonly/APSD_TAI_LAB.git"
IN_COLAB = "google.colab" in sys.modules or "COLAB_RELEASE_TAG" in os.environ

if IN_COLAB:
    # Google Colab: clone the repository (unless already present, e.g. on Drive)
    _repo = Path("/content/APSD_TAI_LAB")
    if not (_repo / "APDS").is_dir():
        subprocess.run(
            ["git", "clone", "--depth", "1", REPO_URL, str(_repo)], check=True
        )
    PROJ_DIR = _repo / "APDS" / PROJECT_NAME
else:
    # Local IDE / Jupyter: find the project folder by walking upward from the
    # working directory (works from the repo root, the project folder or its
    # notebooks/ sub-folder, regardless of where Jupyter was launched).
    _cwd = Path.cwd().resolve()
    PROJ_DIR = next(
        (
            c / "APDS" / PROJECT_NAME
            for c in (_cwd, *_cwd.parents)
            if (c / "APDS" / PROJECT_NAME / ANCHOR_REL).exists()
        ),
        None,
    )
    if PROJ_DIR is None:
        raise FileNotFoundError(
            f"Project folder APDS/{PROJECT_NAME} (anchor: {ANCHOR_REL}) not found "
            f"above {_cwd}. Launch Jupyter from inside the cloned repository."
        )
if str(PROJ_DIR) not in sys.path:
    sys.path.insert(0, str(PROJ_DIR))

print(f"Project folder : {PROJ_DIR}")
print(f"Google Colab   : {IN_COLAB}")
```

## Exercise 2 — 5-Year Heart-Disease Risk Study

A hospital wants to predict, five years ahead, which patients are at risk
of developing heart disease. **Which roles take part in the study, and
what does each contribute?**


```python
# Exercise 2 — role mapping for the heart-disease risk study.
# Each entry: role -> verified responsibility in this study.
ROLES = {
    "Data Scientist": (
        "Frames the clinical question as a predictive problem; explores the"
        " cohort (step 4), engineers features and trains the 5-year risk"
        " model (step 5), evaluates it (step 6)."
    ),
    "Machine Learning Engineer": (
        "Turns the validated model into a deployable service (step 7):"
        " packaging, latency, monitoring hooks for the 5-year horizon."
    ),
    "Data Engineer": (
        "Builds the ingestion pipelines (step 2): hospital records,"
        " lab systems, longitudinal follow-up data; keeps them reliable."
    ),
    "Domain Expert (Cardiologist)": (
        "Supplies clinical knowledge: plausible risk factors, label"
        " definitions, threshold and cost interpretation; validates"
        " findings (steps 1 and 6)."
    ),
    "BI Analyst": (
        "Translates model outputs into dashboards and KPI reports for"
        " hospital management (steps 7-8: deployment of decisions,"
        " monitoring of programme impact)."
    ),
}

for role, contribution in ROLES.items():
    print(f"{role}:")
    print(f"  {contribution}")
```

### Critical thinking

The handout pairs the role list with critical-thinking questions about
the study. The exact wording is `TO_CONFIRM_FROM_TP1_PDF`; the prompts
below stay inside the verified scope (roles + 5-year risk horizon) and
are the notebook's working answers.


```python
CRITICAL_THINKING = [
    {
        "question": (
            "Why does a 5-year risk prediction need a domain expert as"
            " much as a data scientist?"
        ),
        "answer": (
            "Only a cardiologist can judge which variables are clinically"
            " plausible risk factors, how the 'developed heart disease'"
            " label should be defined over 5 years, and whether an"
            " error is a false alarm or a missed patient. Without that,"
            " the model can learn artifacts of the records instead of"
            " physiology."
        ),
    },
    {
        "question": (
            "Why is data engineering central in a hospital study?"
        ),
        "answer": (
            "Patient histories live across silos (admissions, labs,"
            " imaging, follow-up visits). The 5-year horizon demands"
            " longitudinal linkage of records per patient; without clean,"
            " joined pipelines the model trains on the wrong cohorts and"
            " the evaluation is meaningless."
        ),
    },
    {
        "question": (
            "What changes between a one-off study and a deployed 5-year"
            " risk service?"
        ),
        "answer": (
            "Deployment (step 7) and monitoring (step 8) become permanent"
            " work: the ML engineer serves predictions, and the team must"
            " watch for data drift (new recording systems, changing"
            " populations) that silently invalidates a model trained on"
            " today's records."
        ),
    },
]

for item in CRITICAL_THINKING:
    print(f"Q: {item['question']}")
    print(f"A: {item['answer']}")
    print()
```
