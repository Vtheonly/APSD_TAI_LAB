### How each classification is defended

Chapter 1 separates the five fields by the question they answer:

- **Data Science** — statistical pattern extraction and predictive modeling
  on a business problem (present and future horizon; the full 8-step
  workflow).
- **Machine Learning** — designing algorithms that generalize from
  training data (the modeling engine inside a data-science project).
- **Artificial Intelligence** — designing autonomous agents that perceive,
  decide and act (chess, robotics, language); the broadest field.
- **Big Data** — ingesting, storing and transforming high-velocity
  distributed data (Volume, Velocity, Variety).
- **Business Intelligence (BI)** — descriptive and diagnostic analysis of
  past operations (dashboards, KPI reports).

A quick self-check: scenario 1 needs the *whole workflow* on historical
data (Data Science), scenario 2 is about *learning from behavior*
(ML), scenario 3 is about *terabyte-scale infrastructure* (Big Data),
scenario 4 only *describes the past* (BI), and scenario 5 is an
*intelligent agent* (AI).

## Exercise 2 — Roles in a Data Science Project

> **Scenario (handout):** a hospital wants to build a system that predicts
> whether a patient is at risk of heart disease within the next 5 years.
> Several specialists will be involved in this project: **Data Analyst,
> Data Scientist, Data Engineer, ML Engineer, and Business Analyst**.
>
> **Task 1 — Match the Roles.** Assign the following tasks to the correct
> role(s):
>
> - (a) Collect patient records from multiple hospital databases and
>   prepare them for use.
> - (b) Build a predictive model using patient data to estimate heart
>   disease risk.
> - (c) Create a dashboard to show current statistics about patient risk
>   factors (age, weight, smoking).
> - (d) Deploy the predictive model so doctors can access it via a web
>   application.
> - (e) Define the business question: "How can we reduce the number of
>   undetected high-risk patients?"

The mapping below assigns each task to exactly one specialist and
justifies it with that role's responsibility in the 8-step workflow.


```python
# Exercise 2, Task 1 - match each handout task (a)-(e) to one specialist.
ROLE_MATCHING = [
    {
        "task": (
            "(a) Collect patient records from multiple hospital databases"
            " and prepare them for use."
        ),
        "role": "Data Engineer",
        "why": (
            "Designs the ingestion pipelines and integrates the hospital"
            " databases so reliable, analysis-ready data exists"
            " (workflow steps 2-3)."
        ),
    },
    {
        "task": (
            "(b) Build a predictive model using patient data to estimate"
            " heart disease risk."
        ),
        "role": "Data Scientist",
        "why": (
            "Frames the clinical question statistically, engineers"
            " features, trains and evaluates the 5-year risk model"
            " (workflow steps 4-6)."
        ),
    },
    {
        "task": (
            "(c) Create a dashboard to show current statistics about"
            " patient risk factors (age, weight, smoking)."
        ),
        "role": "Data Analyst",
        "why": (
            "Descriptive analysis of existing records: aggregates and"
            " visualizes current risk-factor statistics (SQL,"
            " spreadsheets, dashboards) - no predictive modeling."
        ),
    },
    {
        "task": (
            "(d) Deploy the predictive model so doctors can access it via"
            " a web application."
        ),
        "role": "ML Engineer",
        "why": (
            "Productionizes the validated model: serving, web"
            " integration, latency and drift monitoring"
            " (workflow step 7)."
        ),
    },
    {
        "task": (
            "(e) Define the business question: 'How can we reduce the"
            " number of undetected high-risk patients?'"
        ),
        "role": "Business Analyst",
        "why": (
            "Translates the hospital's strategic concern into the"
            " project's analytical question and success criteria"
            " (workflow step 1)."
        ),
    },
]

for item in ROLE_MATCHING:
    print(item["task"])
    print(f"   Role : {item['role']}")
    print(f"   Why  : {item['why']}")
    print()
```

### Task 2 — Critical Thinking (handout questions)

> **Explain in your own words** why a **Data Engineer** and a **Machine
> Learning Engineer** cannot simply replace each other. Also explain why a
> **Data Analyst** and a **Data Scientist** cannot fully replace each
> other.

The answers below contrast each pair by responsibility, using the role
boundaries of Chapter 1 (`01.2`, *The Boundary Between Roles*).


```python
# Exercise 2, Task 2 - the two handout critical-thinking questions.
CRITICAL_THINKING = [
    {
        "question": (
            "Why can a Data Engineer and a Machine Learning Engineer not"
            " simply replace each other?"
        ),
        "answer": (
            "They work at opposite ends of the pipeline. The Data"
            " Engineer designs and maintains the data infrastructure -"
            " ingestion from the multiple hospital databases, storage,"
            " integration and quality - so that reliable, analysis-ready"
            " data exists at all. The ML Engineer starts once a model"
            " exists and turns it into a production service - serving,"
            " web integration, latency optimization, drift monitoring."
            " Without the Data Engineer there is no trustworthy data to"
            " learn from; without the ML Engineer the validated model"
            " never reaches the doctors."
        ),
    },
    {
        "question": (
            "Why can a Data Analyst and a Data Scientist not fully replace"
            " each other?"
        ),
        "answer": (
            "The Analyst answers descriptive and diagnostic questions -"
            " what happened and why - with SQL, spreadsheets and"
            " dashboards built on existing data. The Scientist answers"
            " predictive and prescriptive questions - what will happen"
            " and what should we do - by engineering features and"
            " training and evaluating statistical and ML models,"
            " including their uncertainty. Removing the Analyst loses"
            " the hospital's clear view of current risk factors (task"
            " c); removing the Scientist loses the 5-year risk"
            " prediction at the heart of the project (task b)."
        ),
    },
]

for item in CRITICAL_THINKING:
    print(f"Q: {item['question']}")
    print(f"A: {item['answer']}")
    print()
```

## Summary

- The five fields of Exercise 1 answer different needs: Data Science runs
